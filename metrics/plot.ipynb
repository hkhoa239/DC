```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
ga = pd.read_csv('ga.csv')
rf = pd.read_csv('rf.csv')
pso = pd.read_csv('pso.csv')
gwo = pd.read_csv('gwo.csv')
```

```python
pga = ga['Power_Consumption']
prf = rf['Power_Consumption']
ppso = pso['Power_Consumption']
pgwo = gwo['Power_Consumption']
```

```python
n = 100  

# Tính tổng reward mỗi n step

# Vẽ biểu đồ
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(pga) + 1), pga, label='Power Consumption GA')
plt.plot(range(1, len(ppso) + 1), ppso, label='Power Consumption PSO')
plt.plot(range(1, len(pgwo) + 1), pgwo, label='Power Consumption GWO')
plt.plot(range(1, len(prf) + 1), prf, label='Power Consumption RF')
plt.xlabel(f'Window (each {n} steps)')
plt.ylabel('Total Power')
plt.title(f'Comparison of Power Consumption per {n} Steps')
plt.legend()
plt.grid()
plt.show()
```

```python
imCpuGA = ga['CPU_Imbalance']
imCPuPSO = pso['CPU_Imbalance']
imCpuGWO = gwo['CPU_Imbalance']
imCpuRF = rf['CPU_Imbalance']
```

```python
# Tính tổng reward mỗi n step

# Vẽ biểu đồ
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(imCpuGA) + 1), imCpuGA, label='CPU Imbalance GA')
plt.plot(range(1, len(imCPuPSO) + 1), imCPuPSO, label='CPU Imbalance PSO')
plt.plot(range(1, len(imCpuGWO) + 1), imCpuGWO, label='CPU Imbalance GWO')
plt.plot(range(1, len(imCpuRF) + 1), imCpuRF, label='CPU Imbalance RF')
plt.xlabel(f'Window (each {n} steps)')
plt.ylabel('CPU Imbalance')
plt.title(f'Comparison of CPU Imbalance per {n} Steps')
plt.legend()
plt.grid()
plt.show()
```
