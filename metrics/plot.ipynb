```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
ga = pd.read_csv('ga.csv')
pso = pd.read_csv('pso.csv')
gwo = pd.read_csv('gwo.csv')
aco = pd.read_csv('aco.csv')
rf = pd.read_csv('rf.csv')
d_ddql = pd.read_csv('d_ddql.csv')
mdql = pd.read_csv('mdql.csv')
nisga = pd.read_csv('nisga.csv')
rd = pd.read_csv('rd.csv')
mf = pd.read_csv('mf.csv')
lf = pd.read_csv('lf.csv')
```

```python
pga = ga['Power_Consumption']
prf = rf['Power_Consumption']
ppso = pso['Power_Consumption']
pgwo = gwo['Power_Consumption']
pd_ddql = d_ddql['Power_Consumption']
paco = aco['Power_Consumption']
pmdql = mdql['Power_Consumption']
pnisga = nisga['Power_Consumption']
prd = rd['Power_Consumption']
pmf = mf['Power_Consumption']
plf = lf['Power_Consumption']
```

```python
n = 100  

# Tính tổng reward mỗi n step

# Vẽ biểu đồ
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(pga) + 1), pga, label='Power Consumption GA')
plt.plot(range(1, len(ppso) + 1), ppso, label='Power Consumption PSO')
plt.plot(range(1, len(pgwo) + 1), pgwo, label='Power Consumption GWO')
plt.plot(range(1, len(prf) + 1), prf, label='Power Consumption RF')
plt.plot(range(1, len(prd) + 1), prd, label='Power Consumption RD')
plt.plot(range(1, len(pmf) + 1), pmf, label='Power Consumption MF')
plt.plot(range(1, len(plf) + 1), plf, label='Power Consumption LF')
plt.plot(range(1, len(pd_ddql) + 1), pd_ddql, label='Power Consumption D_DDQL')
# plt.plot(range(1, len(pmdql) + 1), pmdql, label='Power Consumption MDQL')
plt.plot(range(1, len(paco) + 1), paco, label='Power Consumption ACO')
plt.plot(range(1, len(nisga) + 1), pnisga, label='Power Consumption NISGA')
plt.xlabel(f'Window (each {n} steps)')
plt.ylabel('Total Power')
plt.title(f'Comparison of Power Consumption per {n} Steps')
plt.legend()
plt.grid()
plt.show()
```

```python
imCpuGA = ga['CPU_Imbalance']
imCPuPSO = pso['CPU_Imbalance']
imCpuGWO = gwo['CPU_Imbalance']
imCpuRF = rf['CPU_Imbalance']
imCpuD_ddql = d_ddql['CPU_Imbalance']
imCpuAco = aco['CPU_Imbalance']
imCpuMDql = mdql['CPU_Imbalance']
imCpuNisga = nisga['CPU_Imbalance']
imCpuRD = rd['CPU_Imbalance']
imCpuMF = mf['CPU_Imbalance']
imCpuLF = lf['CPU_Imbalance']
```

```python
# Tính tổng reward mỗi n step

# Vẽ biểu đồ
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(imCpuGA) + 1), imCpuGA, label='CPU Imbalance GA')
plt.plot(range(1, len(imCPuPSO) + 1), imCPuPSO, label='CPU Imbalance PSO')
plt.plot(range(1, len(imCpuGWO) + 1), imCpuGWO, label='CPU Imbalance GWO')
plt.plot(range(1, len(imCpuRF) + 1), imCpuRF, label='CPU Imbalance RF')
plt.plot(range(1, len(imCpuRD) + 1), imCpuRD, label='CPU Imbalance RD')
plt.plot(range(1, len(imCpuMF) + 1), imCpuMF, label='CPU Imbalance MF')
plt.plot(range(1, len(imCpuLF) + 1), imCpuLF, label='CPU Imbalance LF')
plt.plot(range(1, len(imCpuD_ddql) + 1), imCpuD_ddql, label='CPU Imbalance D_DDQL')
plt.plot(range(1, len(imCpuMDql) + 1), imCpuMDql, label='CPU Imbalance MDQL')
plt.plot(range(1, len(imCpuAco) + 1), imCpuAco, label='CPU Imbalance ACO')
plt.plot(range(1, len(imCpuNisga) + 1), imCpuNisga, label='CPU Imbalance NISGA')
plt.xlabel(f'Window (each {n} steps)')
plt.ylabel('CPU Imbalance')
plt.title(f'Comparison of CPU Imbalance per {n} Steps')
plt.legend()
plt.grid()
plt.show()
```

```python
imRamGA = ga['Ram_Imbalance']
imRamPSO = pso['Ram_Imbalance']
imRamGWO = gwo['Ram_Imbalance']
imRamRF = rf['Ram_Imbalance']
imRamD_ddql = d_ddql['Ram_Imbalance']
imRamAco = aco['Ram_Imbalance']
imRamMdql = mdql['Ram_Imbalance']
imRamNisga = nisga['Ram_Imbalance']
imRamRD = rd['Ram_Imbalance']
imRamMF = mf['Ram_Imbalance']
imRamLF = lf['Ram_Imbalance']
```

```python
# Tính tổng reward mỗi n step

# Vẽ biểu đồ
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(imRamGA) + 1), imRamGA, label='Ram Imbalance GA')
plt.plot(range(1, len(imRamPSO) + 1), imRamPSO, label='Ram Imbalance PSO')
plt.plot(range(1, len(imRamGWO) + 1), imRamGWO, label='Ram Imbalance GWO')
plt.plot(range(1, len(imRamRF) + 1), imRamRF, label='Ram Imbalance RF')
plt.plot(range(1, len(imRamRD) + 1), imRamRD, label='Ram Imbalance RD')
plt.plot(range(1, len(imRamMF) + 1), imRamMF, label='Ram Imbalance MF')
plt.plot(range(1, len(imRamLF) + 1), imRamLF, label='Ram Imbalance LF')
plt.plot(range(1, len(imRamD_ddql) + 1), imRamD_ddql, label='Ram Imbalance D_DDQL')
plt.plot(range(1, len(imRamMdql) + 1), imRamMdql, label='Ram Imbalance MDQL')
plt.plot(range(1, len(imRamAco) + 1), imRamAco, label='Ram Imbalance ACO')
plt.plot(range(1, len(imRamNisga) + 1), imRamNisga, label='Ram Imbalance NISGA')
plt.xlabel(f'Window (each {n} steps)')
plt.ylabel('Ram Imbalance')
plt.title(f'Comparison of Ram Imbalance per {n} Steps')
plt.legend()
plt.grid()
plt.show()
```
